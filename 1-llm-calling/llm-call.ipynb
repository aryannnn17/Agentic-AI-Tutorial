Using Google Gemini API

```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
response = llm.invoke("how many states are in India?")
print(response.text)
```

```python
response = llm.invoke([
    ["system","You are a helpful assistant and answer in 6 words"],
    ["user","What is artificial intelligence?"]
])
print(response.content)
```

```python
prompt = "Write a haiku about artificial intelligence"

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
response = llm.invoke(prompt,temperature=0)
print(response.content)
```

```python
response = llm.invoke(prompt,temperature=1)
print(response.content)
```

Using Groq API

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="llama-3.1-8b-instant")
messages = [
    ("system","You are a helpful assistant that answers in rhymes"),
    ("human","What is the capital of France?")
]

response = llm.invoke(messages)
print(response.content)
```
