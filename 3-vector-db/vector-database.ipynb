```python
import chromadb
client = chromadb.Client()  
```

```python
# Delete the collection if it exists
try:
    client.delete_collection(name="documents")
    print("Collection 'documents' deleted")
except:
    print("Collection 'documents' does not exist or could not be deleted")
```

```python
collections = client.create_collection(name="documents", get_or_create=True)
collections.add(
    ids = ["doc1","doc2", "doc3", "doc4"],
    documents = ["ferrari formula 1 car won 16 championships", "mercedes f1 manager is toto wolff", "red bull energy drink gives you wings", "the quick brown fox jumps over the lazy dog",]
)
```

```python
data = collections.get(
    include= ["documents","embeddings","metadatas"]
)

for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument{i+1}")
    print("Text:", data["documents"][i])
    print("Embedding:", len(emb))
    print("First 10 elements of embedding:")
    print(emb[:10])
```

```python
result = collections.query(
    query_texts=["i love doing a lot of sports"],
    n_results=2
)
print(result)
```
