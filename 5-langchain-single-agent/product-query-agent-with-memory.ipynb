# LangChain Single Agent with Memory

```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain.agents import create_agent
load_dotenv()
```

```python
PRODUCTS = {
    "laptop": {
        "name": "MacBook Pro",
        "price": 1299.99,
        "description": "A high-performance laptop for professionals. Chip: M3 Pro. Battery: 10 hours"
    },
    "phone": {
        "name": "iPhone 17",
        "price": 899.99,
        "description": "A premium smartphone with advanced features. Chip: A17 Pro. Battery: 12 hours"
    },
    "airpods": {
        "name": "Airpods Pro",
        "price": 249.99,
        "description": "Wireless earbuds with active noise cancellation. Chip: H2. Battery: 5 hours"
    },
    "watch": {
        "name": "Apple Watch",
        "price": 399.99,
        "description": "A smartwatch with health and fitness tracking. Chip: S9. Battery: 18 hours"
    },
    "charger": {
        "name": "MagSafe Charger",
        "price": 39.99,
        "description": "A wireless charger for iPhone. Chip: None. Battery: N/A"
    }
}
```

```python
@tool
def get_product_info(product_name: str) -> str:
    """Look up product information by name."""
    p = PRODUCTS.get(product_name.lower())
    if not p:
        return f"Product {product_name} not found. Available products: {', '.join(PRODUCTS.keys())}"
    return str(p)
```

```python
llm = ChatGroq(model="llama-3.1-8b-instant", temperature=0)
agent = create_agent(
    llm,
    tools=[get_product_info],
    system_prompt="You are a helpful product assistant. Use the tool to get product information."
)
```

```python
def ask(question: str):
    result = agent.invoke({"messages": [{"role": "user", "content": question}]})
    print(result["messages"][-1].content)
```

```python
ask("What is the price of the iphone?")
```

```python
REVIEWS = {
    "laptop": { 
        "reviews" : 1230,       
        "rating": 4.5
    },
    "phone": {
        "reviews" : 4560,
        "rating": 4.7
    },
    "airpods": {
        "reviews" : 7890,
        "rating": 4.8
    },
    "watch": {
        "reviews" : 2340,
        "rating": 4.6
    },
    "charger": {
        "reviews" : 5670,
        "rating": 4.4
    }
}

@tool
def get_product_reviews(product: str) -> str:
    """Get reviews for a product."""
    r = REVIEWS.get(product.lower())
    if not r:
        return f"Product not found"
    return str(r)
```

```python
llm = ChatGroq(model="llama-3.1-8b-instant", temperature=0)
agent2 = create_agent(
    llm,
    tools=[get_product_info, get_product_reviews],
    system_prompt="You are a helpful product assistant. Use the tool to get product information."
)
```

```python
def ask2(question: str):
    result = agent2.invoke({"messages": [{"role": "user", "content": question}]})
    print(result["messages"][-1].content)
```

```python
ask2("Why people are buying Watch?")
```

```python
ask2("what's the price the of this product?")
```

# LangChain Single Agent with Memory

```python
from langgraph.checkpoint.memory import InMemorySaver

llm = ChatGroq(model="llama-3.1-8b-instant", temperature=0)

agent3 = create_agent(
    llm,
    tools=[get_product_info, get_product_reviews],
    system_prompt="You are a helpful product assistant. Use the tool to get product information.",
    checkpointer=InMemorySaver()
)

def ask3(question: str):
    config = {"configurable": {"thread_id": "user-session-1"}}
    result = agent3.invoke({"messages": [{"role": "user", "content": question}]}, config)
    print(result["messages"][-1].content)
```

```python
ask3("Why people are buying Watch?")
```

```python
ask3("what's the price the of this product?")
```
